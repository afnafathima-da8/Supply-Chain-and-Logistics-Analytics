```python
missing = df.isnull().sum()

missing = missing[missing > 0]

print(missing)
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_percentage = missing_percentage[missing_percentage > 0]

print(missing_percentage.sort_values(ascending=False))
```

```python
date_columns = [
    'order date (DateOrders)',
    'shipping date (DateOrders)'
]

for col in date_columns:
    df[col] = pd.to_datetime(df[col], errors='coerce')

print(df[date_columns].dtypes)
```

```python
print("Dataset Shape:", df.shape)

print("Missing Values:", df.isnull().sum().sum())

print("Duplicate Rows:", df.duplicated().sum())
```

```python
missing_data = pd.DataFrame({
    'Column': df.columns,
    'Missing_Count': df.isnull().sum().values,
    'Missing_Percentage': (df.isnull().sum().values / len(df)) * 100
})

missing_data = missing_data[missing_data['Missing_Count'] > 0]

missing_data = missing_data.sort_values(
    by='Missing_Percentage',
    ascending=False
)

missing_data
```

```python
plt.figure(figsize=(12, 6))

sns.heatmap(df.isnull(), cbar=False)

plt.title('Missing Values in Supply Chain Dataset')
plt.show()
```

```python
high_missing = missing_data[
    missing_data['Missing_Percentage'] > 50
]

high_missing
```

```python
df['Delivery_Delay_Days'] = (
    df['Days for shipping (real)'] -
    df['Days for shipment (scheduled)']
)

df['Delivery_Performance'] = np.where(
    df['Delivery_Delay_Days'] > 0,
    'Late',
    'On Time'
)

df['Delivery_Performance'].value_counts()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x='Delivery_Performance')

plt.title('Delivery Performance Analysis')
plt.xlabel('Delivery Status')
plt.ylabel('Number of Orders')

plt.show()
```

```python
delivery_counts = df['Delivery_Performance'].value_counts()

print(delivery_counts)
```

```python
delivery_percentage = df['Delivery_Performance'].value_counts(normalize=True) * 100

print(delivery_percentage)
```

```python
plt.figure(figsize=(7,5))

plt.pie(
    delivery_counts,
    labels=delivery_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Overall Delivery Performance')
plt.show()
```

```python
shipping_analysis = df.groupby('Shipping Mode')['Delivery_Delay_Days'].mean().sort_values(ascending=False)

print(shipping_analysis)
```

```python
plt.figure(figsize=(10,5))

shipping_analysis.plot(kind='bar')

plt.title('Average Delivery Delay by Shipping Mode')
plt.xlabel('Shipping Mode')
plt.ylabel('Average Delay (Days)')

plt.xticks(rotation=45)
plt.show()
```

```python
region_analysis = df.groupby('Order Region')['Delivery_Delay_Days'].mean().sort_values(ascending=False)

print(region_analysis)
```

```python
plt.figure(figsize=(12,6))

region_analysis.plot(kind='bar')

plt.title('Average Delivery Delay by Region')
plt.xlabel('Order Region')
plt.ylabel('Average Delay (Days)')

plt.xticks(rotation=90)
plt.show()
```

```python
# Convert order date into datetime format
df['order date (DateOrders)'] = pd.to_datetime(
    df['order date (DateOrders)'],
    errors='coerce'
)

# Create month and year columns
df['Order_Month'] = df['order date (DateOrders)'].dt.month_name()

df['Order_Year'] = df['order date (DateOrders)'].dt.year

print("Order_Month and Order_Year created!")
```

```python
monthly_analysis = df.groupby('Order_Month')['Delivery_Delay_Days'].mean()

print(monthly_analysis)
```

```python
month_order = [
    'January', 'February', 'March', 'April',
    'May', 'June', 'July', 'August',
    'September', 'October', 'November', 'December'
]

monthly_analysis = monthly_analysis.reindex(month_order)

print(monthly_analysis)
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    monthly_analysis.index,
    monthly_analysis.values,
    marker='o'
)

plt.title('Monthly Delivery Delay Trends')
plt.xlabel('Month')
plt.ylabel('Average Delivery Delay (Days)')

plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```
